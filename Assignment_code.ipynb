```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix
import seaborn as sns
```

```python
df = pd.read_csv('/content/The_Cancer_data_1500_V2.csv')
```

```python
print('First 5 rows of dataset:')
print(df.head())
```

```python
print('\nDataset Information:')
df.info()
```

```python
print('\nMissing Values:')
print(df.isnull().sum())
```

```python
#Step 2
```

```python
X = df.drop(columns=['Diagnosis']) #Features
y = df['Diagnosis'] #Target variable
```

```python
#Step 3
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print(f'Training samples: {len(X_train)}, Testing samples: {len(X_test)}')
```

```python
#Step 4
```

```python
model = SVC(kernel='linear')
model.fit(X_train, y_train)
print("SVM model training completed successfully.")
```

```python
#Step 5
```

```python
predictions = model.predict(X_test)
percentage = model.score(X_test, y_test)
print(f'Predicted labels: {predictions}')
print(f'Accuracy = {percentage * 100}%')
```

```python
#Step 6
```

```python
import numpy as np
res = confusion_matrix(y_test, predictions)
print("Confusion Matrix:")
print(res)
```

```python
#Step 7
```

```python
nr = pd.DataFrame([[45, 1, 24.5, 1, 2, 3, 5, 0]], columns=X.columns)
prediction_new = model.predict(nr)
print(f"The Prediction for new patient is: {'Cancer' if prediction_new[0] == 1 else 'No Cancer'}")

```
